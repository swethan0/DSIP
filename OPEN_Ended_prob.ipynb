```python
import matplotlib.pyplot as plt
import numpy as np

# Set figure style for clear viewing in Jupyter Notebook
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["font.size"] = 10
```

```python
# Signal 1: Discrete Sequence x[n]
n1 = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 6])
x1 = np.array([2, 3, 1, 5, 2, 3, 1, 3, 2, 3])

plt.figure()
plt.stem(n1, x1, basefmt="k-")
plt.title("Signal 1: Discrete Sequence $x[n]$")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.xticks(n1)
plt.show()
```

```python
# Signal 2: Staircase Waveform
t2 = np.array([0, 1, 1, 2, 2, 3, 3, 4, 4, 5])
x2 = np.array([1, 1, 2, 2, 3, 3, 2, 2, 1, 1])

plt.figure()
plt.plot(t2, x2, "b-", linewidth=2)
plt.title("Signal 2: Staircase Signal")
plt.xlabel("n / t")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```

```python
# Signal 3: Continuous Piecewise Signal
t3 = np.array([-2, 0, 0, 1, 2])
x3 = np.array([-2, -2, -2, 2, 2])

plt.figure()
plt.plot(t3, x3, "r-", linewidth=2)
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.title("Signal 3: Piecewise Signal")
plt.xlabel("n / t")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```

```python
# Signal 4: Shifted Unit Step Combination
n4 = np.arange(-2, 11)
u = lambda n: np.where(n >= 0, 1, 0)
x4 = u(n4) - u(n4 - 3) - 5 * u(n4 - 7)

plt.figure()
plt.stem(n4, x4, basefmt="k-")
plt.title("Signal 4: $x[n] = u(n) - u(n-3) - 5u(n-7)$")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.xticks(n4)
plt.show()
```

```python
# Signal 5: Impulse Combination
n5 = np.arange(-3, 4)
delta = lambda n: np.where(n == 0, 1, 0)
x5 = delta(n5) + 3 * delta(n5 - 1) + 5 * delta(n5 + 1)

plt.figure()
plt.stem(n5, x5, basefmt="k-")
plt.title("Signal 5: $x[n] = \\delta(n) + 3\\delta(n-1) + 5\\delta(n+1)$")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.xticks(n5)
plt.show()
```
