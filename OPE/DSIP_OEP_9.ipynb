```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/image1_noisy_face.png')
kernel = np.ones((5, 5), np.float32) / 25
result = cv2.filter2D(image, -1, kernel)

print("Averaging Filter")
cv2_imshow(result)
cv2.imwrite('/content/image1_averaging.jpg', result)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-09-04 154435.png')

# Convert to float to avoid overflow
image_float = image.astype(np.float32)

laplacian = cv2.Laplacian(image_float, cv2.CV_32F)

result = image_float - 0.5 * laplacian

# Keep pixel values in valid range
result = np.clip(result, 0, 255).astype(np.uint8)

print("Laplacian Sharpened")
cv2_imshow(result)

cv2.imwrite('/content/image5_laplacian.png', result)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-09-04 154435.png')

# High-pass sharpening kernel
kernel = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
], dtype=np.float32)

image_float = image.astype(np.float32)

result = cv2.filter2D(image_float, -1, kernel)

# Normalize result so it displays correctly
result = cv2.normalize(result, None, 0, 255, cv2.NORM_MINMAX)

result = result.astype(np.uint8)

print("High-Pass Sharpened")
cv2_imshow(result)

cv2.imwrite('/content/image5_highpass.png', result)
```
