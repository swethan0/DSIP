```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-3, 7)
x = np.array([2, 3, 2, 3, 2, 3, 2, 3, 2, 3])

print("n =", n)
print("x[n] =", x)

plt.figure(figsize=(10, 5))
plt.stem(n, x)
plt.title("Discrete-Time Signal x[n]")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.xticks(n)
plt.yticks([0, 1, 2, 3, 4])
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 6)
x = np.array([1, 2, 3, 3, 2, 1])

print("n =", n)
print("x[n] =", x)

plt.figure(figsize=(10, 5))

plt.stem(n, x)

plt.title("Discrete-Time Staircase Signal x[n]")
plt.xlabel("n")
plt.ylabel("x[n]")

plt.xticks(n)
plt.yticks([0, 1, 2, 3])

plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.array([-2, -1, 0, 1, 2])
x = np.array([-2, -2, 0, 2, 0])

print("n =", n)
print("x[n] =", x)

plt.figure(figsize=(10, 5))

plt.stem(n, x)

plt.title("Discrete-Time Ramp Signal x[n]")
plt.xlabel("n")
plt.ylabel("x[n]")

plt.xticks(n)
plt.yticks([-2, -1, 0, 1, 2, 3])

plt.axhline(0, linewidth=1)

plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-3, 11)

u = lambda n: np.where(n >= 0, 1, 0)

x = u(n) - u(n - 3) - 5 * u(n - 7)

print("n =", n)
print("x[n] =", x)

plt.figure(figsize=(10, 5))

plt.stem(n, x)

plt.title(r"Discrete-Time Signal $x[n]=u[n]-u[n-3]-5u[n-7]$")
plt.xlabel("n")
plt.ylabel("x[n]")

plt.xticks(n)
plt.yticks([-5, -4, -3, -2, -1, 0, 1])

plt.axhline(0, linewidth=1)

plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 6)

delta = lambda n: np.where(n == 0, 1, 0)

x = delta(n) + 3 * delta(n - 1) + 5 * delta(n + 1)

print("n =", n)
print("x[n] =", x)

plt.figure(figsize=(10, 5))

plt.stem(n, x)

plt.title(r"Discrete-Time Signal $x[n]=\delta[n]+3\delta[n-1]+5\delta[n+1]$")
plt.xlabel("n")
plt.ylabel("x[n]")

plt.xticks(n)
plt.grid(True)

plt.show()
```
