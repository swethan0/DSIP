```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(
    r"C:\Users\admin\Documents\DSIP\Lab Experiment\dark_car.png",
    cv2.IMREAD_GRAYSCALE
)

if image is None:
    raise FileNotFoundError("Image not found. Check the image path.")

plt.figure(figsize=(8, 6))
plt.imshow(image, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

```python
_, binary = cv2.threshold(
    image,
    127,
    1,
    cv2.THRESH_BINARY
)

plt.figure(figsize=(8, 6))
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")
plt.show()
```

```python
pattern = np.array([
    [-1,  1, -1],
    [ 1,  1,  1],
    [-1,  1, -1]
], dtype=np.int8)

print("Structuring Element:")
print(pattern)
```

```python
result = cv2.morphologyEx(
    binary,
    cv2.MORPH_HITMISS,
    pattern
)

print("Hit-or-Miss Transformation Completed!")
```

```python
plt.figure(figsize=(8, 6))
plt.imshow(result * 255, cmap="gray")
plt.title("Hit-or-Miss Transformation")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(result * 255, cmap="gray")
plt.title("Hit-or-Miss Output")
plt.axis("off")

plt.tight_layout()
plt.show()
```
