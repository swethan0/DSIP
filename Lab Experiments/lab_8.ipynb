```python
import numpy as np
import matplotlib.pyplot as plt 
import cv2
```

```python
def generate_histogram(image):

    histogram = cv2.calcHist(
        [image],
        [0],
        None,
        [256],
        [0, 256]
    ) 
    return histogram
```

```python
input_path = (r"C:\study\codes\DSIP\low_con.jpg")
input_image = cv2.imread(input_path)


gray_image = cv2.cvtColor(input_image, cv2.COLOR_BGR2GRAY)

input_histogram = generate_histogram(gray_image)

intensity_values = np.arange(256)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)    
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.bar(intensity_values, input_histogram, color='blue')
plt.title('Image Histogram')
plt.xlabel('Intensity Value')
plt.ylabel('Frequency')
plt.xlim([0, 255])
plt.grid(alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
total_pixels = gray_image.size
print("Total number of pixels in the image:", total_pixels)
normalized_histogram = input_histogram / total_pixels

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.bar(intensity_values, normalized_histogram, color='black')
plt.title('Normalized Histogram')
plt.xlabel('Intensity Value')
plt.ylabel('Normalized Frequency')
plt.xlim([0, 255])
```
